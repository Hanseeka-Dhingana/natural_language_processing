```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
sentences = ['I love my dog. I love my cat.']

# instance of tokenizer
# tokenize for 100 of words
tokenizer = Tokenizer(num_words=100, oov_token="<OOV>")

tokenizer.fit_on_texts(sentences)

word_index = tokenizer.word_index 

word_index
```

OOV stands for Out-Of-Vocabulary.

An OOV token (like "<OOV>" or "[UNK]") is a special token used to represent words that the model hasn’t seen before—words not in the vocabulary the tokenizer was trained 

If we update the sentences list by adding the new sentence with exclamation symbol. Tokenizer don't create the new tokenizer for this it combine with the word where exclamation is used. It only create the new token for you. 

```python
sentences = ['I love my dog.', 'I love my cat.', 'you love my dog! and also cat']

tokenizer.fit_on_texts(sentences)

word_index = tokenizer.word_index 

print(word_index)
```

**Convert Text_to_Sequences**    

A sequence is a list of numbers that represent words or tokens from a text. These numbers are just IDs from a vocabulary.

```python
sequences = tokenizer.texts_to_sequences(sentences)

sequences
```

pad_sequences is a utility function (e.g., from Keras) that makes sure all sequences (lists of token IDs) have the same length by:
- Padding shorter sequences with zeros (or a specified value)
- Or truncating longer sequences

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

padded = pad_sequences(sequences)

print(padded)
```

```python
padded  = pad_sequences(sequences, padding="post", maxlen=6)
print(padded)
```
